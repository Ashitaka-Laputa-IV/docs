# 安装 LangGraph

安装 LangGraph 基础包:

```bash pip
pip install -U langgraph
```

```bash uv
uv add langgraph
```

```bash npm
npm install @langchain/langgraph @langchain/core
```

```bash pnpm
pnpm add @langchain/langgraph @langchain/core
```

```bash yarn
yarn add @langchain/langgraph @langchain/core
```

```bash bun
bun add @langchain/langgraph @langchain/core
```

使用 LangGraph 时, 你通常需要访问 LLM 并定义 tools。
你可以按自己认为合适的方式来实现。

其中一种方式(也是本文档中采用的方式)是使用 [LangChain](https://docs.langchain.com/oss/langchain/overview)。

安装 LangChain:

```bash pip
pip install -U langchain
# Requires Python 3.10+
```

```bash uv
uv add langchain
# Requires Python 3.10+
```

```bash npm
npm install langchain
```

```bash pnpm
pnpm add langchain
```

```bash yarn
yarn add langchain
```

```bash bun
bun add langchain
```

要使用特定 LLM provider 的包, 你需要单独安装。

各 provider 的具体安装说明请参阅 [integrations](https://docs.langchain.com/oss/integrations/providers/overview) 页面。
