# LangGraph 概览

从 Klarna、Uber、J.P. Morgan 到更多正在塑造 agent 未来的公司都信赖 LangGraph。LangGraph 是一个底层编排 framework 与 runtime, 用于构建、管理和部署长时运行、有状态的 agent。LangGraph 让你能进行细粒度控制, 在同一个 graph 中把确定性的手工编写步骤与 LLM 驱动的 agentic 步骤混合起来, 从而构建出行为完全符合你的应用要求的定制 agent。

LangGraph 非常底层, 且完全专注于 agent **编排**。在使用 LangGraph 之前, 我们建议你先熟悉一些用于构建 agent 的组件, 从 [models](https://docs.langchain.com/oss/langchain/models) 与 [tools](https://docs.langchain.com/oss/langchain/tools) 开始。

我们在整套文档中通常会使用 [LangChain](https://docs.langchain.com/oss/langchain/overview) 组件来集成 models 与 tools, 但你并不需要使用 LangChain 才能使用 LangGraph。如果你刚开始接触 agent, 或者想要更高层的抽象, 我们建议你使用 LangChain 的 [agents](https://docs.langchain.com/oss/langchain/agents), 它为常见的 LLM 与 tool-calling loop 提供了 prebuilt 架构。

LangGraph 专注于 agent 编排所依赖的底层能力: durable execution、streaming、human-in-the-loop 等等。

LangGraph 的核心优势之一, 是能在单个 graph 中把确定性步骤与 LLM 驱动的 agentic 步骤混合起来。这让你可以构建定制 workflow: 其中一部分逻辑完全可预测、可审计, 而另一部分则灵活且由 model 驱动, 从而让你精细控制 AI 究竟在何处、以何种方式被应用。

**LangChain 各产品如何配合**

- [Deep Agents](https://docs.langchain.com/oss/deepagents/overview) 是一个 [agent harness](https://docs.langchain.com/oss/concepts/products#agent-harnesses-like-the-deep-agents-sdk): 在 LangGraph 之上提供 planning、subagents、filesystem tools 与上下文管理。
- [LangChain](https://docs.langchain.com/oss/langchain/overview) 是 agent framework: 为 models、tools 与 agent loop 提供抽象与集成。
- [LangGraph](https://docs.langchain.com/oss/langgraph/overview) 是编排 runtime: durable execution、streaming、human-in-the-loop 与 persistence。
- [LangSmith](https://docs.langchain.com/langsmith/observability) 是用于跨 framework 追踪、evaluation、prompt 与部署的平台。
- [LangSmith Engine](https://docs.langchain.com/langsmith/engine) 能检测你的 LangGraph agent trace 中的问题并提出修复方案。你可以直接在 Engine 标签页中打开一个包含所提修复的 pull request。
- [LangSmith Fleet](https://docs.langchain.com/langsmith/fleet/index) 是 no-code agent 构建器, 用于模板、集成与例行自动化。

阅读 [framework、runtime 与 harness 对比](https://docs.langchain.com/oss/concepts/products) 了解开源技术栈的比较。

````python
# 环境准备: 从 .env 加载 API keys
from dotenv import load_dotenv

load_dotenv()  # 自动向上查找仓库根目录的 .env

# ---- 统一导入 ----
from langgraph.graph import StateGraph, MessagesState, START, END
````

````python
from IPython.display import display, Image
from langgraph.graph.state import CompiledStateGraph

def show_graph(graph: CompiledStateGraph) -> None:
    display(Image(graph.get_graph().draw_mermaid_png()))

from rich import print as rprint
````

然后, 创建一个简单的 hello world 示例:

````python
def mock_llm(state: MessagesState):
    return {"messages": [{"role": "ai", "content": "hello world"}]}


graph = StateGraph(MessagesState)
graph.add_node(mock_llm)
graph.add_edge(START, "mock_llm")
graph.add_edge("mock_llm", END)
graph = graph.compile()

show_graph(graph)

res = graph.invoke({"messages": [{"role": "user", "content": "hi!"}]})

rprint(res)
````

````python
res = graph.invoke({"messages": [{"role": "user", "content": "hello!"}]})

rprint(res)
````

由此可以体现, 对于简单的图, 消息并没有累加, 而只是HUMNA+AI一对消息. 

## 核心优势

LangGraph 为*任何*长时运行、有状态的 workflow 或 agent 提供底层支撑基础设施。LangGraph 不对 prompt 或架构做抽象, 并提供以下核心优势:

* **混合确定性步骤与 agentic 步骤**: 在单个 graph 中把手工编写的确定性逻辑与 LLM 驱动的决策结合起来。在需要可靠性与可预测性的地方使用确定性步骤, 在需要灵活性的地方使用 agentic 步骤, 从而让你精确控制 agent 行为的每一部分。
* [Persistence](https://docs.langchain.com/oss/langgraph/persistence): 构建即使遭遇故障也能继续、可长时间运行的 agent, 并从上次中断处恢复执行。
* [Human-in-the-loop](https://docs.langchain.com/oss/langgraph/interrupts): 通过在任何时刻检查并修改 agent state 引入人工监督。
* [全面的 memory](https://docs.langchain.com/oss/concepts/memory): 创建有状态的 agent, 既具备用于持续推理的 short-term working memory, 也具备跨会话的 long-term memory。
* [使用 LangSmith 调试](https://docs.langchain.com/langsmith/observability): 借助可视化工具深入洞察复杂的 agent 行为, 这些工具能追踪执行路径、捕获 state 转换并提供详细的 runtime 指标。
* [生产就绪的部署](https://docs.langchain.com/langsmith/deployment): 借助可扩展的基础设施自信地部署复杂的 agent 系统, 从容应对有状态、长时运行 workflow 带来的独特挑战。

## LangGraph 生态

LangGraph 可以单独使用, 也能与任何 LangChain 产品无缝集成, 为开发者提供构建 agent 的完整工具套件。要改进你的 LLM 应用开发, 可以把 LangGraph 与以下产品搭配使用:

- [**LangSmith Observability**](https://docs.langchain.com/langsmith/observability): 在一处追踪请求、评估输出并监控部署。用 LangGraph 在本地做原型, 再借助集成的 observability 与 evaluation 走向生产, 构建更可靠的 agent 系统。
- [**LangSmith Deployment**](https://docs.langchain.com/langsmith/deployment): 借助专为长时运行、有状态 workflow 打造的部署平台, 轻松部署并扩展 agent。跨团队发现、复用、配置并共享 agent, 还可以通过 Studio 中的可视化原型快速迭代。
- [**LangChain**](https://docs.langchain.com/oss/langchain/overview): 提供集成与可组合组件, 简化 LLM 应用开发。其中包含构建在 LangGraph 之上的 agent 抽象。
